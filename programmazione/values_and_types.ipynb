# Programmi


Quando programmiamo, implementiamo algoritmi attraverso una sequenza di `istruzioni`, che risulta in un programma. Le istruzioni seguono una *sintassi* precisa, che ci indica *come* scrivere correttamente una istruzione. Seguono anche una *semantica*, che indica cosa ogni istruzione fa.
L'interprete poi le esegue per noi, una di sequito all'altra. 


Nota: quando lavoriamo su Pycharm (o su Colab) alcune parti delle istruzioni vengono colorate automaticamente per aiutarci a leggere meglio il codice.

````python
from Bio import SeqIO
from Bio.Blast.NCBIWWW import qblast


# legge un dataset salvato in data/flora.gbk
flora = list(SeqIO.parse("data/flora.gbk", "genbank"))

# filtra: seleziona solo dati relativi alle orchidee
orchids = list(filter(lambda flower: "orchid" in flower.annotations["references"][0].title, flora))

# estrae le sequenze genetiche
sequences = list(map(lambda orchid: orchid.seq, orchids))

# ricerca sequenze simili
if len(orchids) < 3:
    similar_sequences = [qblast("blastn", "nt", sequence) for sequence in sequences]
````

## Variabili: linee guida


- Usa nomi significativi, e.g. `hypothenuse` al posto di `hp`, `genome` al posto di `gnm`
- Non usare nomi eccessivamente corti quando possibile, e.g., `n`, `x`
- Per nomi con spazi, usa `_`, e.g., `complement_genome`
- Non iniziare con numeri, e.g., `2genome`, e usali quanto meno possibile
- Usa il minuscolo
- Per costanti, e.g., `GRAVITY`, usa maiuscoletto

### Espressioni su tipi collezione

- `element in collection` l'elemento e' nella collezione?
- `collection[indice]` accedi all'elemento associato all'indice `indice` nella collezione
- `collection[:indice]` sottocollezione che parte dall'inizio fino all'indice `indice`
- `collection[indice:]` sottocollezione che parte dall'indice `indice` e arriva alla fine
- `collection[inizio:fine]` sottocollezione che parte dall'indice `inizio` e arriva all'indice `fine`
